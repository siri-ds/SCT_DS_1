```python
import pandas as pd
import matplotlib.pyplot as plt
import zipfile
```

```python
zip_file = "API_SP.POP.TOTL_DS2_en_csv_v2_267553.zip"

with zipfile.ZipFile(zip_file, 'r') as z:
    csv_file = [f for f in z.namelist() if f.endswith('.csv') and 'Metadata' not in f][0]

    with z.open(csv_file) as file:
        df = pd.read_csv(file, skiprows=4)
```

```python
df.head()
```

```python
population = df[['Country Name', '2024']]
population.head()
```

```python
population = population.dropna()
```

```python
top10 = population.sort_values(by='2024', ascending=False).head(10)

top10
```

```python
print(top10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(top10['Country Name'], top10['2024'])

plt.title('Top 10 Most Populated Countries')
plt.xlabel('Country')
plt.ylabel('Population')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
population_values = population['2024']

plt.figure(figsize=(10,6))
plt.hist(population_values, bins=20)

plt.title('Distribution of Population')
plt.xlabel('Population')
plt.ylabel('Number of Countries')

plt.tight_layout()
plt.show()
```
